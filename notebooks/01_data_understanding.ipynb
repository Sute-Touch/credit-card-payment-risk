```python
id_checks = pd.Series({
    "Количество строк": len(df),
    "Уникальных ID": df["ID"].nunique(),
    "Пропусков ID": df["ID"].isna().sum(),
    "Повторяющихся ID": df["ID"].duplicated().sum(),
    "Полных дубликатов строк": df.duplicated().sum()
})

id_checks
```

```python
category_columns = [
    "SEX",
    "EDUCATION",
    "MARRIAGE",
    "PAY_0",
    "PAY_2",
    "PAY_3",
    "PAY_4",
    "PAY_5",
    "PAY_6"
]

for column in category_columns:
    print(f"\n{column}")
    print(df[column].value_counts(dropna=False).sort_index())
```

```python
numeric_columns = [
    "LIMIT_BAL",
    "AGE",
    *[f"BILL_AMT{i}" for i in range(1, 7)],
    *[f"PAY_AMT{i}" for i in range(1, 7)]
]

numeric_summary = df[numeric_columns].describe(
    percentiles=[0.01, 0.50, 0.99]
).T

numeric_summary.round(2)
```

```python
negative_counts = (
    df[numeric_columns]
    .lt(0)
    .sum()
    .rename("Количество отрицательных значений")
)

negative_counts
```

```python
from sklearn.model_selection import train_test_split

TARGET = "default payment next month"
RANDOM_STATE = 42

train_df, test_df = train_test_split(
    df,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=df[TARGET]
)

train_df = train_df.copy()
test_df = test_df.copy()
```

```python
split_summary = pd.DataFrame({
    "Выборка": ["Обучающая", "Тестовая"],
    "Количество клиентов": [len(train_df), len(test_df)],
    "Количество событий": [
        train_df[TARGET].sum(),
        test_df[TARGET].sum()
    ],
    "Доля событий, %": [
        train_df[TARGET].mean() * 100,
        test_df[TARGET].mean() * 100
    ]
})

split_summary.round(2)
```

```python
overlap = set(train_df["ID"]) & set(test_df["ID"])

len(overlap)
```

```python
from pathlib import Path

output_dir = Path("project_data")
output_dir.mkdir(exist_ok=True)

train_df.to_csv(output_dir / "train.csv", index=False)
test_df.to_csv(output_dir / "test.csv", index=False)

split_ids = pd.concat([
    train_df[["ID"]].assign(sample="train"),
    test_df[["ID"]].assign(sample="test")
], ignore_index=True)

split_ids.to_csv(output_dir / "split_ids.csv", index=False)
```
